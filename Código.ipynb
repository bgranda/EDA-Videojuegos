## Comparación de archivos

Parece que tenemos archivos iguales, por lo que vamos a proceder a compararlos antes de unirlos.

```python
if df2.equals(df4):
    print("Los DataFrames df2 y df4 son iguales.")
else:
    print("Los DataFrames df2 y df4 son diferentes.")
```

```python
if df3.equals(df5):
    print("Los DataFrames df3 y df5 son iguales.")
else:
    print("Los DataFrames df3 y df5 son diferentes.")
```

## Procederemos a unir solo los dataframes distintos, ya que df2 y df4 son iguales.

```python
dataframes = [df1_reordered, df2, df3, df5, df7_reordered]
df_combined = pd.concat(dataframes, axis=0, ignore_index=True)
df_combined.info()
```

```python
df_combined['Year'] = df_combined['Year'].astype('Int64')
```

```python
df_limpio = df_combined.drop_duplicates(keep='first', ignore_index=True)
df_limpio.info()
```

## Visualizacion de valores nulos

```python
filas_nulas = df_limpio[df_limpio['Game'].isnull()]
filas_nulas
```

```python
# Los valores nulos que se encuentran en la columna 'Game' son problemáticos, ya que esta columna es esencial para identificar los videojuegos. 
# Por lo tanto, es recomendable eliminar estas filas del DataFrame para asegurar la integridad de los datos.
df_limpio.dropna(subset=['Game'], inplace=True)
df_limpio.info()
```

```python
df_limpio.to_csv('Videojuegos.csv', index=False, encoding='utf-8')
```

## Comparación de archivos

Parece que tenemos archivos iguales, por lo que vamos a proceder a compararlos antes de unirlos.

```python
if df2.equals(df4):
    print("Los DataFrames df2 y df4 son iguales.")
else:
    print("Los DataFrames df2 y df4 son diferentes.")
```

```python
if df3.equals(df5):
    print("Los DataFrames df3 y df5 son iguales.")
else:
    print("Los DataFrames df3 y df5 son diferentes.")
```

## Procederemos a unir solo los dataframes distintos, ya que df2 y df4 son iguales.

```python
dataframes = [df1_reordered, df2, df3, df5, df7_reordered]
df_combined = pd.concat(dataframes, axis=0, ignore_index=True)
df_combined.info()
```

```python
df_combined['Year'] = df_combined['Year'].astype('Int64')
```

```python
df_limpio = df_combined.drop_duplicates(keep='first', ignore_index=True)
df_limpio.info()
```

## Visualizacion de valores nulos

```python
filas_nulas = df_limpio[df_limpio['Game'].isnull()]
filas_nulas
```

```python
# Los valores nulos que se encuentran en la columna 'Game' son problemáticos, ya que esta columna es esencial para identificar los videojuegos. 
# Por lo tanto, es recomendable eliminar estas filas del DataFrame para asegurar la integridad de los datos.
df_limpio.dropna(subset=['Game'], inplace=True)
df_limpio.info()
```

## 4. ANÁLISIS DE DATOS

```python
# Análisis de ventas globales por plataforma de videojuegos
analisis_ventas_plataforma = df_limpio.groupby('Platform')['Global'].describe()
analisis_ventas_plataforma
```

```python
plt.figure(figsize=(10, 6))
plt.bar(analisis_ventas_plataforma.index, analisis_ventas_plataforma.values, color='skyblue')
plt.title('Ventas Globales por Plataforma de Videojuegos', fontsize=16)
plt.xlabel('Plataforma', fontsize=12)
plt.ylabel('Ventas Globales', fontsize=12)
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
# Analisis de ventas globales por género de videojuegos
analisis_ventas_genero = df_limpio.groupby('Genre')['Global'].describe()
analisis_ventas_genero
```

```python
plt.figure(figsize=(10, 6))
plt.bar(analisis_ventas_genero.index, analisis_ventas_genero.values, color='lightgreen')
plt.title('Ventas Globales por Género de Videojuegos', fontsize=16)
plt.xlabel('Género', fontsize=12)
plt.ylabel('Ventas Globales', fontsize=12)
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
# Análisis de ventas en función de la plataforma y el género de los videojuegos
ventas_plataforma_genero = df_limpio.groupby(['Platform', 'Genre'])['Global'].describe()
ventas_plataforma_genero
```

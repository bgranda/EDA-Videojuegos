```python
# Definir una función para que haga el treemap en función de géneros y región para poder hacer las gráficas.

def treemap_generos_region(df, region, year_start, year_end):
''' Definir la variable sales en función del año, año que empieza y año que termina, se agrupa por Género, coges la columna de la región que quieres'''
    sales = (
        df[
            (df['Year'] >= year_start) &
            (df['Year'] <= year_end)
        ]
        .groupby('Genre')[region]
        .sum()
        .reset_index()
    )

    fig = px.treemap(
        sales,
        path=['Genre'],
        values=region,
        color=region,
        color_continuous_scale='Viridis'
    )

    fig.update_traces(
        textinfo='label+percent root'
    )

    fig.update_layout(
        title=f'{region} ({year_start}-{year_end})'
    )

    fig.show()
```

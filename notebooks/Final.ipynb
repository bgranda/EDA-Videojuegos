# EDA DE VENTAS DE VIDEOJUEGOS

En este Análisis de Mercado de las Ventas de Videojuegos se estudian los datos desde 1980 a 2020.

En este análisis, se ha podido concluir lo siguiente:

- Con el cambio de siglo, hay un incremento de las ventas de videojuegos, debido a la evolución de las consolas.
- También, se puede concluir, que con el cambio de siglo, tanto a nivel global como por regiones, cambian los géneros más vendidos a nivel global y también por regiones.
- Las ventas de videojuegos están muy influenciadas porque un mismo videojuego se vende en para diferentes consolas.
- En el top 20 de las ventas de Videojuegos, podemos destacar que los más vendidos son Wii Sports y Grand Theft Auto V junto con la franquicia de Call of Duty.
- En el análisis de los videojuegos más vendidos por regiones, podemos concluir que Europa y Norte América venden principalmente Wii Sports y Grand Theft Auto V. Mientras que en Japón, prefieren las ventas están centradas en las franquicias de Pokemon, Animal Crossing y Super Mario Bros.
- También hemos podido concluir, que la venta de videojuegos no es uniforme sino que el 80% de las ventas se concentra en un sólo un 16% de videojuegos.
- A nivel global las ventas de videojuegos se concentran principalmente para las siguientes consolas: PS3, X360 y Wii. Sin embargo, en el análisis por regiones se puede destacar que Norte América tiene destacan las ventas por las consolas X360 y Wii, y en el resto de las regiones estudiadas, están distribuidas de forma pareja entre todas las consolas.
- Pero para saber quién es realmente líder en el sector de las ventas de videojuegos por consolas, las hemos agrupado por fabricante. Los líderes a nivel global y por regiones son claramente: Nintendo y Sony.
- A nivel global, el género más vendido es el de Acción, seguido por Deportes y Disparos. Mientras que al estudiarlos por regiones, Norte América y Europa mantienen la misma tendencia global, pero Japón el género más vendido es el de Rol.
- Si estudiamos las ventas de videojuegos según su género y el fabricante de las consolas, centrándonos en los tres principales: Sony y Nintendo. Se concluye, que cada fabricante es líder en distintos genéros según su ventas, por lo que se puede concluir que los públicos objetivos de las consolas son distintos. Sony se centra principalmente en el género de Acción, y luego en Deportes y Disparos. Sin embargo, Nintendo se centra de forma más o menos igualada en Arcade, Simuladores de Conducción, Deportes y Acción.

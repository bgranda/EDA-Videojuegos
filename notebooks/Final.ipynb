# EDA DE VENTAS DE VIDEOJUEGOS

En este Análisis de Mercado de las Ventas de Videojuegos se estudian los datos desde 1980 a 2020.

HIPÓTESIS:

1) Cómo han variado las ventas de videojuegos a lo largo de los años
2) Videojuegos más vendidos y preferencias por regiones
3) Consolas más populares
4) Qué fabricante de consolas es el lider en la venta de videojuegos
5) Género de videojuego líder en el sector a nivel global y su variabilidad por regiones
6) Preferencias de género de videojuegos por fabricante de consolas, ¿varía el público objetivo?

CONCLUSIONES:

- Con el cambio de siglo, hay un incremento de las ventas de videojuegos, debido a la evolución de las consolas, y la conectividad entre las mismas. Debido a esto, también cambiaron las preferencias de géneros más vendidas hasta el momento.

- De los videojuegos más vendidos a nivel mundial, podemos destacar que los más vendidos son Wii Sports y Grand Theft Auto V junto con la franquicia de Call of Duty.En el análisis de los videojuegos más vendidos por regiones, podemos concluir que Europa y Norte América venden principalmente Wii Sports y Grand Theft Auto V. Mientras que en Japón, prefieren las ventas están centradas en las franquicias de Pokemon, Animal Crossing y Super Mario Bros.

- Las ventas de videojuegos están muy influenciadas porque un mismo videojuego se vende en para diferentes consolas.
- Además, que también están muy influenciados por los superventas, ya que el 16% de los videojuegos ocupan el 80% de las ventas a nivel mundial.

- Las consolas más populares a nivel global son: PS3, X360 y Wii. Sin embargo, en el análisis por regiones se puede destacar que Norte América tiene destacan las ventas por las consolas X360 y Wii, y en el resto de las regiones estudiadas, están distribuidas de forma pareja entre todas las consolas.

- Los fabricantes de consolas líderes en la venta de videojuegos son Sony y Nintendo, dejando muy atrás a Microsoft.

- Las preferencias de género de videojuegos a nivel global son los de Acción. Sin embargo, al estudiarlos por regiones, mientras que en Norte América y Europa mantienen las mismas preferencias que a nivel global, Japón difiere y tiene como preferencia los juegos de Rol.


- Si estudiamos las ventas de videojuegos según su género y el fabricante de las consolas, centrándonos en los dos principales: Sony y Nintendo.  Sony se centra principalmente en el género de Acción, y luego en Deportes y Disparos. Sin embargo, Nintendo se centra de forma más o menos igualada en Arcade, Simuladores de Conducción, Deportes y Acción. Después del análisis realizado anteriormente sobre los juegos más vendidos, se puede ver que aunque a pesar de que algunos géneros coincidad en ambos fabricantes, los videojuegos son totalmente diferentes. De ahí se puede concluir, que cada fabricante es líder en distintos genéros según su ventas y en los géneros que coincidan, ya que sus grupos objetivos de usuarios son distintos.

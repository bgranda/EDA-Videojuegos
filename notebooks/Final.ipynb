# EDA DE VENTAS DE VIDEOJUEGOS

En este Análisis de Mercado de las Ventas de Videojuegos se estudian los datos desde 1980 a 2020.

HIPÓTESIS:

1) Cómo han variado las ventas de videojuegos a lo largo de los años
2) Videojuegos más vendidos y preferencias por regiones
3) Consolas más populares
4) Qué fabricante de consolas es el lider en la venta de videojuegos
5) Género de videojuego líder en el sector a nivel global y su variabilidad por regiones
6) Preferencias de género de videojuegos por fabricante de consolas, ¿varía el público objetivo?

CONCLUSIONES:

- Con el cambio de siglo, hay un incremento de las ventas de videojuegos, debido a la evolución de las consolas, y la conectividad entre las mismas. Debido a esto, también cambiaron las preferencias de géneros más vendidas hasta el momento.

- De los videojuegos más vendidos a nivel mundial, podemos destacar que los más vendidos son Wii Sports y Grand Theft Auto V junto con la franquicia de Call of Duty.En el análisis de los videojuegos más vendidos por regiones, podemos concluir que Europa y Norte América venden principalmente Wii Sports y Grand Theft Auto V. Mientras que en Japón, prefieren las ventas están centradas en las franquicias de Pokemon, Animal Crossing y Super Mario Bros.

- Las ventas de videojuegos están muy influenciadas porque un mismo videojuego se vende en para diferentes consolas.
- Además, que también están muy influenciados por los superventas, ya que el 16% de los videojuegos ocupan el 80% de las ventas a nivel mundial.

- Las consolas más populares a nivel global son: PS3, X360 y Wii. Sin embargo, en el análisis por regiones se puede destacar que Norte América tiene destacan las ventas por las consolas X360 y Wii, y en el resto de las regiones estudiadas, están distribuidas de forma pareja entre todas las consolas.

- Los fabricantes de consolas líderes en la venta de videojuegos son Sony y Nintendo, dejando muy atrás a Microsoft.

- Las preferencias de género de videojuegos a nivel global son los de Acción. Sin embargo, al estudiarlos por regiones, mientras que en Norte América y Europa mantienen las mismas preferencias que a nivel global, Japón difiere y tiene como preferencia los juegos de Rol.


- Si estudiamos las ventas de videojuegos según su género y el fabricante de las consolas, centrándonos en los dos principales: Sony y Nintendo.  Sony se centra principalmente en el género de Acción, y luego en Deportes y Disparos. Sin embargo, Nintendo se centra de forma más o menos igualada en Arcade, Simuladores de Conducción, Deportes y Acción. Después del análisis realizado anteriormente sobre los juegos más vendidos, se puede ver que aunque a pesar de que algunos géneros coincidad en ambos fabricantes, los videojuegos son totalmente diferentes. De ahí se puede concluir, que cada fabricante es líder en distintos genéros según su ventas y en los géneros que coincidan, ya que sus grupos objetivos de usuarios son distintos.

TITULO EDA: LA HISTORIA DE LOS VIDEOJUEGOS: DE LOS 8 BITS AL 4K

En este proyecto voy a contar la historia de los videojuegos: desde los 8 bits a los videojuegos 4K, a través de las ventas de los videojuegos a lo largo de los años.
El primer análisis de evolución de los videojuegos a lo largo de los años se dividen en dos épocas: 1980 - 1999 y 2000 - 2020. Esto es debido, a la gran evolución
de las consolas y el desarrollo del internet y fibra óptica, y los juegos colaborativos que hubo en el cambio de siglo. Por este motivo, hay un gran incremento de 
ventas de videojuegos a partir del año 2000, un cambio de géneros de los videojuegos más vendidos y la aparición de nuevos géneros de videojuegos y 
desaparición de otros.

Los géneros más vendidos de 1980 a 1999 son principalmente Arcade, y después los de Rol y Acción. Sin embargo, los más vendidos del 2000 a 2020 son principalmente
los de Acción, y después los de Deportes y Disparos.

En los 80s y 90s, podemos ver que Norte América y Europa tienen un comportamiento parecido en las ventas de videojuegos por género siendo Arcade y Acción. 
Mientras que en Japón dominaba los juegos de Rol y Arcade. Las ventas en el Resto del Mundo se centraban en Arcade y Rol. En las décadas de los 2000s, 
se ve un cambio en las ventas por género por regiones. Norte América y Europa tienen las mismas tendencias: Acción, Deportes y Disparos. Mientras que Japón, 
sigue manteniendo las ventas en los juegos de Rol mayoritariamente, y después Acción. En el Resto del Mundo, también hay un cambio de ventas en el Resto del Mundo 
cambiando a los géneros de Acción y Deportes.

Después de ver la evolución de los géneros de los videojuegos a lo largo de los años, vamos a estudiar cuáles son los videojuegos más vendidos. En el análisis, lo
primero que he hecho ha sido realizar un análisis descriptivo, y después de normalizar los datos para el top ten de las ventas, se pudo concluir de que las ventas
de los videojuegos están influenciados porque un mismo videojuego se vende para distintas consolas. Por ello, ajustamos nuestro análisis para que sólo tenga
los videojuegos que son diferentes.

Aunque a nivel global, el videojuego más vendido es el Wii Sports, seguido de Gran Theft Auto V. En este top de los 20 videojuegos más vendidos, se puede ver que
 la franquicia Call of Duty es la más vendida. Con este último análisis, podemos concluir que los gustos de Norte América y Europa son más similares, ya que 
 coinciden en los videojuegos de Wii Sports y la franquicia de Call of Duty. Aunque en Norte América, les gusta más los juegos de la franquicia de Super Mario Bros 
 y en Europa prefieren la franquicia de la FIFA. Sin embargo, en Japon prefieren la franquicia de Pokemon, Animal Crossing y Super Mario Bros diferiendo mucho de 
 los gustos de Norte América y Europa.

Una vez estudiados los videojuegos más populares, nos planteamos si las ventas de los mismos se concentran en unos pocos superventas o no. Y el resultado de este 
análisis es que el 80% de las ventas globales se concentran sólo en un 16% de los videojuegos, es decir, los superventas.

Una vez visto los videojuegos más populares, vamos a ver cuáles son las consolas preferidas en el mundo. Las consolas más utilizadas a nivel mundial son: PS3, X360 y Wii. Sin embargo, al estudiar las consolas más populares por regiones podemos ver: que en Norte América las más populares son la X360, la Wii y la DS. Sin embargo,
en las demás regiones, las consolas populares están distribuidas más equitativamente por todas las consolas del momento.

Una vez visto esto, podríamos pensar que las consolas PS3, X360 y Wii son líderes en el sector. Pero, la verdad es que conviven muchas consolas al mismo tiempo de 
un mismo fabricante. Por lo que, lo que tendremos que estudiar realmente es qué fabricante de consolas es realmente líder en el sector en la venta de videojuegos.
Al hacer este análisis, vemos que claramente hay dos fabricantes líderes en este sector tanto a nivel global como por regiones: Sony y Nintendo, dejando a Microsoft
en tercer lugar pero bastante alejado.

Una vez visto los análisis anteriores, hemos querido analizar las ventas de viedojuegos según el género y el fabricante de consolas, para saber si variaban mucho
y si van dirigidas al mismo público objetivo o no. Sony tiene un porcentaje muy alto de sus ventas en los juegos de Acción, seguidos por Deportes, Disparos, y luego por Rol y Carreras. Mientras que en Nintendo, tiene una distribución de géneros más equitativa entre los de Arcade, Simulador de Conducción, Deportes, Acción y Rol.
Por lo que podemos observar que al diferir tanto en los géneros, podemos concluir que los públicos objetivos de ambos fabricantes son distintos por lo que no compiten realmente en sus ventas.

Al hacer, este análisis por regiones obtenemos las siguientes conclusiones: Analizando las gráficas por regiones, en Norte América se puede visualizar que el género más vendido es el de Acción y que el fabricante que lo lidera es Sony. El siguiente género más vendido es el de disparos por Microsoft. Después le siguen los géneros Simuladores de Conducción, Arcade y Deportes. Los dos primeros están liderados por Nintendo, mientras que Deportes está muy igualado en ventas tanto Sony como Nintendo.

En Europa, el género más vendido también es Acción por el fabricante de Sony. Le siguen Disparos y Deportes también de Sony, y por último Arcade y Simulador de conducción por Nintendo.

Ahora podemos ver un cambio radical en la venta en Japón, ya que los juegos más vendidos son los de Rol de Nintendo. Luego le siguen los de Arcade de Nintendo, y de unas ventas similares son los Simuladores de Conducción de Nintendo y los de Acción de Sony.

En el resto del mundo podemos ver un comportamiento similar a Europa y Norte América, liderando el sector el género de Acción por el fabricante Sony, y después de forma muy igualada Disparos y Deportes también por parte de Sony.
